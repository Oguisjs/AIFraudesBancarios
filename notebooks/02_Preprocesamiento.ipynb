```python
# CELDA 1: Importar módulos
import sys
sys.path.append('../src')
from preprocessing import *
import numpy as np
import os
```

```python
# CELDA 2: Correr el pipeline completo
RUTA = '../data/raw/PS_20174392719_1491204439457_log.csv'

df = cargar_datos(RUTA, muestra=1_000_000)
df = ingenieria_caracteristicas(df)
X, y, features = preparar_features(df)
X_train, X_val, X_test, y_train, y_val, y_test = dividir_y_balancear(X, y)
X_train_sc, X_val_sc, X_test_sc, scaler = escalar(X_train, X_val, X_test)
```
