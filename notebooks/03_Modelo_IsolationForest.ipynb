# Modelo 1: Isolation Forest
Entrenamiento del modelo Isolation Forest no supervisado.

```python
# CELDA 1: Cargar librerías y datos
import numpy as np
from sklearn.ensemble import IsolationForest
from sklearn.metrics import classification_report, roc_auc_score
import joblib

X_train = np.load('../data/processed/X_train.npy')
X_test  = np.load('../data/processed/X_test.npy')
y_test  = np.load('../data/processed/y_test.npy')

print('Train size:', X_train.shape)
print('Test size:', X_test.shape)
```

```python
# CELDA 2: Entrenar el modelo
modelo_if = IsolationForest(
    n_estimators=200,
    contamination=0.001,
    max_features=0.75,
    random_state=42,
    n_jobs=-1
)
modelo_if.fit(X_train)
print('Isolation Forest entrenado.')
```

```python
# CELDA 3: Evaluar preliminarmente
pred_raw = modelo_if.predict(X_test)
y_pred_if = np.where(pred_raw == -1, 1, 0)
scores_if = -modelo_if.score_samples(X_test)

print(classification_report(y_test, y_pred_if, target_names=['Legítima','Fraude']))
print(f'AUC-ROC: {roc_auc_score(y_test, scores_if):.4f}')
```

```python
# CELDA 4: Guardar el modelo
import os
os.makedirs('../models', exist_ok=True)
joblib.dump(modelo_if, '../models/isolation_forest.pkl')
print('Modelo guardado: models/isolation_forest.pkl')
```
